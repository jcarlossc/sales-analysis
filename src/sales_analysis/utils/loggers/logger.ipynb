```python
from pathlib import Path
import sys
import import_ipynb
from pyprojroot import here
from sales_analysis.utils.settings.setting_yaml import *
from loguru import logger
```

```python
def configure_logger():
    """
    Configura sistema global de logs do projeto.

    A função cria automaticamente o diretório de logs
    na raiz do projeto e adiciona dois destinos
    de logging:

    1. Terminal (stdout)
    2. Arquivo de log

    Returns
    -------
    loguru.Logger
        Instância global configurada do logger.
    """

    # =-------------------------------------------------
    # Carrega configurações centralizadas do projeto
    # =-------------------------------------------------
    config_yaml = load_config()

    # =-------------------------------------------------
    # Localiza automaticamente a raiz do projeto
    # =-------------------------------------------------
    ROOT = here()

    # =-------------------------------------------------
    # Define caminho do arquivo de configuração
    # =-------------------------------------------------
    LOG_DIR = Path(ROOT / config_yaml["paths"]["logs"])

    # =-------------------------------------------------
    # Diretório de logs na raiz
    # =-------------------------------------------------
    LOG_DIR = Path(ROOT / "logs")

    # =-------------------------------------------------
    # Evita erro caso diretório já exista
    # =-------------------------------------------------
    LOG_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    # =-------------------------------------------------
    # Caminho do arquivo de log
    # =-------------------------------------------------
    log_file = (
        LOG_DIR / config_yaml["logging"]["filename"]
    )

    # =-------------------------------------------------
    # Remove configuração padrão do Loguru e
    # Evita duplicidade de logs no terminal
    # =-------------------------------------------------
    logger.remove()

    # ------------------------------------------
    # Logger arquivo
    # ------------------------------------------
    logger.add(
        log_file,

        level = config_yaml["logging"]["level"],

        format = config_yaml["logging"]["format"]
    )

    logger.info("Logger configurado com sucesso.")

    return logger
    
logger = configure_logger()
```
