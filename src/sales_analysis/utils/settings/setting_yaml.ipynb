```python
# Bibliotecas utilizadas
from pathlib import Path
import yaml
from pyprojroot import here
```

```python
def load_config(verbose: bool = False) -> dict:
    """
    Carrega o arquivo de configuração YAML do projeto.

    A função localiza automaticamente a raiz do projeto
    utilizando ``pyprojroot.here()`` e realiza a leitura
    do arquivo ``config/config.yaml``.

    Returns
    -------
    dict
        Dicionário contendo todas as configurações
        definidas no arquivo YAML.

    Raises
    ------
    FileNotFoundError
        Caso o arquivo de configuração não exista.

    yaml.YAMLError
        Caso o arquivo YAML esteja inválido.

    Examples
    --------
    >>> config = load_config()
    >>> config["paths"]["raw_data"]
    'data/raw/'

    Notes
    -----
    Estrutura esperada do projeto:

    project_root/
    ├── config/
    │   └── config.yaml
    ├── notebooks/
    ├── src/
    └── pyproject.toml

    O uso de ``pyprojroot`` evita dependência de
    caminhos relativos como:

    >>> Path.cwd().parent.parent

    tornando o projeto mais robusto e portátil.
    """

    if verbose:
        print("[INFO] Iniciando carregamento config")

    try:
    
        # =-------------------------------------------------
        # Localiza automaticamente a raiz do projeto
        # =-------------------------------------------------
        ROOT = here()
    
        # =-------------------------------------------------
        # Define caminho do arquivo de configuração
        # =-------------------------------------------------
        CONFIG_PATH = Path(ROOT / "config/config.yaml")
    
        # =-------------------------------------------------
        # Validação do arquivo de configuração
        # =-------------------------------------------------
        if not CONFIG_PATH.exists():
    
            raise FileNotFoundError(
                f"Arquivo de configuração não encontrado: "
                f"{CONFIG_PATH}"
            )
    
        # =-------------------------------------------------
        # Leitura do arquivo YAML
        # =-------------------------------------------------
        with open(CONFIG_PATH, "r", encoding="utf-8") as file:
            config_yaml = yaml.safe_load(file)
    
        return config_yaml

    except yaml.YAMLError as error:
        if verbose:
            print(
                f"[ERROR] YAML inválido: "
                f"{error}"
            )

        raise

    except Exception as error:
        if verbose:
            print(
                f"[CRITICAL] "
                f"Erro inesperado: "
                f"{error}"
            )

        raise

    finally:
        if verbose:
            print(
                "[INFO] "
                "Finalizando load_config"
            )

config_yaml = load_config()
```
