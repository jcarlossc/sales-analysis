```python

def load_csv(
    file_path: str | Path,
    sep: str = ",",
    encoding: str = "utf-8"
) -> pd.DataFrame:
    """
    Realiza leitura de arquivo CSV com tratamento de erros.

    Parameters
    ----------
    file_path : str | Path
        Caminho do arquivo CSV.

    sep : str, optional
        Separador do arquivo CSV.

    encoding : str, optional
        Codificação do arquivo.

    Returns
    -------
    pd.DataFrame
        DataFrame carregado.

    Raises
    ------
    FileNotFoundError
        Caso o arquivo não exista.

    pd.errors.EmptyDataError
        Caso o arquivo esteja vazio.

    pd.errors.ParserError
        Caso o CSV esteja corrompido.

    Exception
        Outros erros inesperados.
    """

    try:
        file_path = Path(file_path)

        logger.info(f"Iniciando leitura do arquivo: {file_path}")

        # Verifica se o arquivo existe
        if not file_path.exists():
            raise FileNotFoundError(
                f"Arquivo não encontrado: {file_path}"
            )

        # Leitura do CSV
        df = pd.read_csv(
            file_path,
            sep=sep,
            encoding=encoding
        )

        # Verifica se dataframe está vazio
        if df.empty:
            logger.warning("O arquivo CSV está vazio.")

        logger.info(
            f"Arquivo carregado com sucesso. "
            f"Linhas: {df.shape[0]} | Colunas: {df.shape[1]}"
        )

        return df

    except FileNotFoundError as error:
        logger.error(error)
        raise

    except pd.errors.EmptyDataError:
        logger.error("O arquivo CSV está vazio.")
        raise

    except pd.errors.ParserError:
        logger.error("Erro ao interpretar o arquivo CSV.")
        raise

    except UnicodeDecodeError:
        logger.error(
            "Erro de codificação. "
            "Tente outro encoding."
        )
        raise

    except Exception as error:
        logger.exception(
            f"Erro inesperado ao carregar CSV: {error}"
        )
        raise 
```
