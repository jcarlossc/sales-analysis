```python
from pathlib import Path
#import logging
import pandas as pd
```

```python
def load_csv(
    file_path: str | Path,
    sep: str = ",",
    encoding: str = "utf-8"
) -> pd.DataFrame:
 
    # Leitura do CSV
    df = pd.read_csv(
        file_path,
        sep=sep,
        encoding=encoding
    )

    return df    
```
