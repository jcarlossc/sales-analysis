```python
def get_graphic_discount_margin(df: pd.DataFrame) -> None:
    """
    Gera gráficos de dispersão para avaliar a relação
    entre descontos aplicados e margens de lucro.

    O objetivo é comparar o comportamento das margens
    considerando dois cenários:

    - Margem calculada com desconto aplicado
    - Margem calculada sem desconto aplicado

    Parameters
    ----------
    df : pd.DataFrame
        DataFrame contendo métricas comerciais.

    Required Columns
    ----------------
    Discount
    margem_com_desc
    margem_sem_desc

    Returns
    -------
    None
        Exibe gráficos comparativos utilizando matplotlib.

    Notes
    -----
    Essa análise auxilia na investigação de:

    - Impacto dos descontos sobre rentabilidade
    - Possíveis perdas de margem
    - Existência de padrões ou correlações
    - Identificação de outliers

    Interpretation
    --------------
    - Tendência descendente:
      descontos maiores → menor margem

    - Tendência horizontal:
      pouco impacto dos descontos

    - Grande dispersão:
      outros fatores influenciam a margem
    """
    
    logger.info(f"Início da geração do gráfico de Disconto vs Margem")
    
    fig, axes = plt.subplots(1, 2, figsize = (14, 5))

    # Com desconto
    axes[0].scatter(df["Discount"], df["margem_com_desc"], alpha = 0.5)
    axes[0].set_title("Margem COM Desconto")
    axes[0].set_xlabel("Desconto")
    axes[0].set_ylabel("Margem")
    axes[0].grid(True)
    
    # Sem desconto
    axes[1].scatter(df["Discount"], df["margem_sem_desc"], alpha = 0.5)
    axes[1].set_title("Margem SEM Desconto")
    axes[1].set_xlabel("Desconto")
    axes[1].set_ylabel("Margem")
    axes[1].grid(True)
    
    plt.tight_layout()
    plt.show()

    logger.info(f"Término da geração do gráfico de Disconto vs Margem")
```
