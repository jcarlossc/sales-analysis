```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import import_ipynb
```

```python
from sales_analysis.utils.loggers.logger import *
```

```python
def get_graphic_impact_discount(df: pd.DataFrame) -> None:
    """
    Simula cenários de desconto e avalia seu impacto
    sobre faturamento, lucro e margem.

    A função aplica diferentes percentuais de desconto
    aos preços unitários e recalcula indicadores
    financeiros para cada cenário.

    Parameters
    ----------
    df : pd.DataFrame
        DataFrame contendo dados comerciais.

    Required Columns
    ----------------
    Unit_Price
    Unit_Cost
    Quantity_Sold

    Returns
    -------
    None
        Exibe gráfico mostrando o comportamento
        do lucro conforme o desconto aumenta.

    Notes
    -----
    Essa análise auxilia na avaliação de:

    - Sensibilidade do lucro aos descontos
    - Faixa de desconto sustentável
    - Identificação do ponto de prejuízo
    - Planejamento comercial

    Assumptions
    -----------
    - Quantidade vendida permanece constante
    - Custos unitários permanecem constantes
    - Apenas o preço é alterado
    """

    logger.info(f"Início da geração do gráfico do desconto sobre o lucro")
    
    # cenários de desconto
    cenarios = np.arange(0, 0.31, 0.02)   # de 0% até 30%
    
    resultados = []
    
    for desconto in cenarios:
        preco_com_desc = df['Unit_Price'] * (1 - desconto)
        faturamento = (preco_com_desc * df['Quantity_Sold']).sum()
        custo = (df['Unit_Cost'] * df['Quantity_Sold']).sum()
        lucro = faturamento - custo
        margem = lucro / faturamento if faturamento != 0 else 0
    
        resultados.append([desconto, faturamento, lucro, margem])
    
    simulacao = pd.DataFrame(
        resultados,
        columns=['desconto', 'faturamento', 'lucro', 'margem']
    )
    
    plt.figure(figsize=(14,6))
    
    plt.plot(simulacao['desconto'] * 100,
             simulacao['lucro'],
             marker='o')
    
    plt.axhline(0, linestyle='--')
    
    plt.title('Impacto do desconto no lucro')
    plt.xlabel('Desconto (%)')
    plt.ylabel('Lucro')
    
    plt.show()

    logger.info(f"Término da geração do gráfico do desconto sobre o lucro")  
```
