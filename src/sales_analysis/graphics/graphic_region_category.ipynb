```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import import_ipynb
```

```python
from sales_analysis.utils.loggers.logger import *
```

```python
def get_graphic_region_category(df: pd.DataFrame) -> None:
    """
    Gera gráficos comparativos de lucro/prejuízo por
    região e categoria de produto, considerando cenários
    com e sem desconto.

    O objetivo é avaliar como os descontos impactam
    a rentabilidade em diferentes segmentos do negócio.

    Parameters
    ----------
    df : pd.DataFrame
        DataFrame contendo informações comerciais.

    Required Columns
    ----------------
    Region
    Product_Category
    lucro_com_desc
    lucro_sem_desc

    Returns
    -------
    None
        Exibe gráficos comparativos utilizando matplotlib.

    Notes
    -----
    Esta análise auxilia na identificação de:

    - Regiões mais rentáveis
    - Categorias com maior impacto dos descontos
    - Segmentos com prejuízo operacional
    - Diferenças entre cenários com e sem desconto
    """

    logger.info(
        "Iniciando gráfico "
        "região x categoria"
    )

    try:

        # ------------------
        # Validações
        # ------------------
        if df.empty:

            logger.warning(
                "DataFrame vazio"
            )

            raise ValueError(
                "DataFrame não pode estar vazio"
            )

        logger.debug(
            "Validações concluídas"
        )

        # ------------------
        # Agregações
        # ------------------
        logger.info(
            "Calculando agregações"
        )
    
        # Agrupamento
        df_region = df.groupby("Region")[["lucro_com_desc", "lucro_sem_desc"]].sum().reset_index()
        df_category = df.groupby("Product_Category")[["lucro_com_desc", "lucro_sem_desc"]].sum().reset_index()
        
        # Configuração
        fig, axes = plt.subplots(1, 2, figsize = (16, 6))
        
        width = 0.35
        
        # Gráfico 1 - Região
        x1 = np.arange(len(df_region))
        
        axes[0].bar(x1 - width/2, df_region['lucro_com_desc'], width, label = 'Com Desconto', color = "orange")
        axes[0].bar(x1 + width/2, df_region['lucro_sem_desc'], width, label = 'Sem Desconto', color = 'blue')
        
        axes[0].axhline(0)
        axes[0].set_title("Lucro/Prejuízo por Região")
        axes[0].set_xticks(x1)
        axes[0].set_xticklabels(df_region["Region"])
        axes[0].legend()
        
        # Gráfico 2 - Categoria
        x2 = np.arange(len(df_category))
        
        axes[1].bar(x2 - width/2, df_category["lucro_com_desc"], width, label="Com Desconto", color = "orange")
        axes[1].bar(x2 + width/2, df_category["lucro_sem_desc"], width, label="Sem Desconto", color = "blue")
        
        axes[1].axhline(0)
        axes[1].set_title("Lucro/Prejuízo por Categoria")
        axes[1].set_xticks(x2)
        axes[1].set_xticklabels(df_category["Product_Category"])
        axes[1].legend()
        
        # Ajuste
        plt.tight_layout()
        plt.show()

        logger.info("Gráfico gerado")

    except KeyError as error:
        logger.error(
            "Erro estrutural: %s",
            error,
            exc_info=True
        )

        raise

    except (
        ValueError,
        TypeError

    ) as error:
        logger.error(
            "Erro validação: %s",
            error,
            exc_info=True
        )

        raise

    except Exception as error:
        logger.critical(
            "Erro inesperado: %s",
            error,
            exc_info=True
        )

        raise

    finally:
        logger.info("Processamento encerrado")
```
