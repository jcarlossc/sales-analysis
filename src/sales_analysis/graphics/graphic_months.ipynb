```python
def get_graphic_months(df: pd.DataFrame) -> None:
    """
    Gera gráficos temporais mensais comparando indicadores
    financeiros com e sem aplicação de descontos.

    O gráfico apresenta a evolução mensal de:

    - Faturamento
    - Custos
    - Lucro

    em dois cenários:

    - Com desconto aplicado
    - Sem desconto aplicado

    Parameters
    ----------
    df : pd.DataFrame
        DataFrame contendo dados comerciais processados.

    Required Columns
    ----------------
    Sale_Date
    faturamento_com_desc
    faturamento_sem_desc
    custo
    lucro_com_desc
    lucro_sem_desc

    Returns
    -------
    None
        Exibe gráficos comparativos utilizando matplotlib.

    Notes
    -----
    A análise mensal auxilia na identificação de:

    - Tendências de vendas
    - Impacto dos descontos
    - Evolução de rentabilidade
    - Sazonalidade do negócio
    """

    logger.info(f"Início da geração do gráfico graphic_months")
    
    df["mes"] = df["Sale_Date"].dt.to_period('M')

    mensal_com_desc = df.groupby("mes").agg({
        "faturamento_com_desc": "sum",
        "custo": "sum",
        "lucro_com_desc": "sum"
    })
    
    mensal_sem_desc = df.groupby("mes").agg({
        "faturamento_sem_desc": "sum",
        "custo": "sum",
        "lucro_sem_desc": "sum"
    })
    
    fig, axes = plt.subplots(1, 2, figsize = (16, 6))  # 1 linha, 2 colunas
    
    # Com desconto
    mensal_com_desc.plot(ax = axes[0])
    axes[0].set_title("Com Desconto")
    axes[0].set_xlabel("Mês")
    axes[0].set_ylabel("Valor")
    axes[0].tick_params(axis="x", rotation = 45)
    
    # Sem desconto
    mensal_sem_desc.plot(ax = axes[1])
    axes[1].set_title("Sem Desconto")
    axes[1].set_xlabel("Mês")
    axes[1].set_ylabel("Valor")
    axes[1].tick_params(axis = "x", rotation = 45)
    
    plt.tight_layout()
    plt.show()

    logger.info(f"Término da geração do gráfico graphic_months")
```
